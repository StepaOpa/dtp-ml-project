```python
import geopandas as gpd
import pandas as pd
import json
from shapely.geometry import Point
import numpy as np
```

```python
file_path = "C:/Users/lenovo/Desktop/URFU/DataAnalysis_3sem/DTP_GIBDD/data/sverdlovskaia-oblast.geojson"
with open(file_path, "r", encoding="utf-8") as f:
    data = json.load(f)

features = data['features']
df = pd.json_normalize(features)
```

Проеобразование координат в тип данных shapely.geometry.Point, проеобразование ячеек без значений в тип даных None, т.к. geopandas отказывается создавать geometry по NoneType значениям

```python
def make_point(coords):
    if coords is None:
        return None
    if isinstance(coords, list) and len(coords) == 2:
        try:
            lon, lat = float(coords[0]), float(coords[1])
            if -180 <= lon <= 180 and -90 <= lat <= 90:
                return Point(lon, lat)
        except (ValueError, TypeError):
            pass
    return None

geometry = df['geometry.coordinates'].apply(make_point)
```

```python
gdf = gpd.GeoDataFrame(df, crs='epsg:4326', geometry=geometry)
gdf.sample(10)
```

```python
def list_to_nan(x):
    if isinstance(x, list) and len(x) == 0:
        return np.nan
    else:
        return x
    
for col in gdf.columns:
    gdf[col] = gdf[col].apply(list_to_nan)
```

```python
gdf.sample(5)
```

```python
gdf['properties.datetime'] = pd.to_datetime(gdf['properties.datetime'])

gdf = gdf[['properties.datetime'] + [col for col in gdf.columns if col not in ['geometry.coordinates', 'properties.datetime'] + ['geometry.coordinates']]]

gdf_sorted = gdf.sort_values('properties.datetime')
gdf_sorted
```
